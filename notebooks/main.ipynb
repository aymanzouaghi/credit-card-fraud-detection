```python
X = df.drop("Class",axis=1)
y=df['Class']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,random_state=42,stratify=y)
```

```python
df["Class"].isna().sum() #no null values in the target variable so we can proceed with model building. without any imputation.
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler =StandardScaler()
```

```python
X_train_scaled =scaler.fit_transform(X_train)
X_test_scaled =scaler.fit_transform(X_test)
```

```python
df['Class'].value_counts()
```

```python
from imblearn.under_sampling import RandomUnderSampler
rus = RandomUnderSampler(random_state=42)
X_train_balnced, y_train_balanced = rus.fit_resample(X_train_scaled, y_train)
```

```python
y_train_balanced.value_counts()
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=1000)
model.fit(X_train_balnced, y_train_balanced)
```

```python
y_pred = model.predict(X_test_scaled)
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

```python
from sklearn.metrics import ( accuracy_score, precision_score, recall_score, f1_score)
print("Accuracy : ", accuracy_score(y_test, y_pred))
print("precision : ", precision_score(y_test, y_pred))
print("recall : ", recall_score(y_test, y_pred))
print("F1 : ", f1_score(y_test, y_pred))
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics  import roc_auc_score
y_prob = model.predict_proba(X_test_scaled)[:, 1]
print("ROC-AUC : ", roc_auc_score(y_test, y_prob))
```
